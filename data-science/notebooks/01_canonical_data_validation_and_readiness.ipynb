```python
import json
from pathlib import Path
import pandas as pd

ROOT = Path('..').resolve() if Path.cwd().name == 'notebooks' else Path('.').resolve()
if (ROOT / 'data-science').exists() is False and (ROOT.parent / 'data').exists():
    ROOT = ROOT.parent

CANONICAL = ROOT / 'data' / 'canonical' / 'intelligence_data.csv'
df = pd.read_csv(CANONICAL)
print(f'Loaded {len(df)} rows x {len(df.columns)} columns | version={df["data_version"].iloc[0]}')
df.head(3)
```

```python
print('--- Record Type Distribution ---')
print(df['record_type'].value_counts())
print('\n--- Source Distribution ---')
print(df['source_name'].value_counts())
print('\n--- Metric & Unit Consistency ---')
print(df.groupby(['record_type', 'metric_name', 'metric_unit'])['metric_value'].agg(['count', 'min', 'median', 'max']))
```

```python
dates = pd.to_datetime(df['observed_at'], utc=True)
print('Date range:', dates.min(), 'to', dates.max())
print('Years by record_type:')
print(pd.crosstab(dates.dt.year, df['record_type']))
print('\n--- Category Distribution ---')
print(df['category'].value_counts())
```
